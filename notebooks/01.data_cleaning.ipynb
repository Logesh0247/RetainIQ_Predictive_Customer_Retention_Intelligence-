```python
df = pd.read_csv(r"../Data/Telco_customer_churn_Dataset.csv")

df.head()
```

```python
df.isnull().sum()
```

## Remove useless column

```python
df.drop("Column1", axis=1, inplace=True)
df.drop("Count", axis=1, inplace=True)
df.drop("Lat Long", axis=1, inplace=True)
```

```python
df.columns
```

## Data Type conversion

Total charges

```python
df["Total Charges"] = pd.to_numeric(df["Total Charges"],errors="coerce")

df["Total Charges"] = (df["Total Charges"].fillna(df["Total Charges"].median()))
```

Convert Zip Code

        --> zip code is converted to str because ZIP code is an identifier.
        --> You never: add ZIP codes, average ZIP codes, multiply ZIP codes.

```python
df["Zip Code"] = (df["Zip Code"].astype(str))
```

## Convert Categories
--> repeated values 

```python
categorical_cols = ["Country","State","City","Gender","Senior Citizen","Partner","Dependents",
                    "Phone Service","Multiple Lines","Internet Service","Online Security","Online Backup",
                    "Device Protection","Tech Support","Streaming TV","Streaming Movies","Contract",
                    "Paperless Billing","Payment Method","Churn Label","Churn Reason"]

for col in categorical_cols:
    df[col] = df[col].astype("category")
```

## Optimizing integer coloumns

```python
# Tenure Months.   
df["Tenure Months"] = (df["Tenure Months"].astype("int16"))

# Churn Value. 
df["Churn Value"] = df["Churn Value"].astype("int8")

# Churn Score. 
df["Churn Score"] = (df["Churn Score"].astype("int16"))

# CLTV
df["CLTV"] = (df["CLTV"].astype("int32"))
```

## Optimize Float Columns

```python
# Latitude
df["Latitude"] = (df["Latitude"].astype("float32"))

# longitude
df["Longitude"] = (df["Longitude"].astype("float32"))

# Monthly charges
df["Monthly Charges"] = (df["Monthly Charges"].astype("float32"))

# Total charges
df["Total Charges"] = (df["Total Charges"].astype("float32"))
```

## Handle Churn Reason

```python
df["Churn Reason"] = df["Churn Reason"].cat.add_categories(["Not Churned"])

df["Churn Reason"] = df["Churn Reason"].fillna("Not Churned")
```

## Remove duplicates

```python
df.drop_duplicates(inplace=True)
```

## Text cleaning

```python
object_cols = (df.select_dtypes(include="object").columns)

for col in object_cols:df[col] = (df[col].str.strip())
```

## Validation

```python
print("Missing Values:",df.isnull().sum().sum())

print("Shape:",df.shape)
```
